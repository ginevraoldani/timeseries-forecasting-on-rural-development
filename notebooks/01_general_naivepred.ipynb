### <a id='toc1_1_3_'></a>[Utility functions](#toc0_)

```python
def historical_mean(df, col_name):
    """
    Applica il modello Naive "Historical Mean":
    La previsione è la media di tutti i valori storici osservati nel train set.
    
    Args:
        df (pd.DataFrame): DataFrame con DatetimeIndex e colonne rinominate.
        col_name (str): Il nome della variabile (colonna) da predire.
        
    Returns:
        tuple: (train_df, test_df) dove test_df contiene la colonna 'pred_mean'.
            Ritorna (None, None) se i dati sono insufficienti.
    """
    series = df[col_name].dropna()
    
    if series.empty:
        print(f"Skipping {col_name}: no data")
        return None, None
    
    n_samples = len(series)
    split_idx = int(n_samples * 0.8)
    
    # Check split valido
    if split_idx == 0 or split_idx >= n_samples:
        print(f"Skipping {col_name}: not enough data for split (n={n_samples})")
        return None, None
    
    train = series.iloc[:split_idx].to_frame(name='Value')
    test = series.iloc[split_idx:].to_frame(name='Value')

    hist_mean = train['Value'].mean()
    
    test['pred_mean'] = hist_mean
    
    print(f"Model: Historical Mean | Var: {col_name} | Train size: {len(train)} | Mean: {hist_mean:.4f}")
    
    return train, test
```

```python
results = {}

for col in df.columns:
    train, test = historical_mean(df, col)
    
    if train is not None:
        results[col] = (train, test)
        metrics = eval.compute_errors(test['Value'].values, test['pred_mean'].values)
        display(metrics)
        print(f"RMSE (test {len(test)} values): {metrics['RMSE']}")
        print(f"MAE (test {len(test)} values): {metrics['MAE']}")
        print(f"MAPE (test {len(test)} values): {metrics['MAPE']:.2f}%")
        #rep.save_metrics(model_name="HISTMEAN", variable_name=col, metrics=metrics)
        visual.plot_forecast(
        train=train,
        test=test,
        variable_name='population_percent',
        model_name='HISTMEAN',
        baseline=test['pred_mean'],
        baseline_name='HISTMEAN',
        prediction=None,
        rmse=metrics['RMSE'],
        save_plot=True
    )
```
